## torch.randn
가우시안 정규분포로 평균이 0, 표준편차가 1인 가우시안 정규분포
np.random.normal과 같이 평균과 표준편차를 정할 수는 없지만 곱하고 더하는식으로 가능
(torch.randn * std) + mean 과 같은 형식으로 진행

## torch.rand

## torch.linspace
---
`torch.linspace(start, end, step)` 으로 진행되며 시작점과 끝점 사이를 지정한 개수(step)만큼 균등한 간격으로 나눈 값들을 텐서로 생성한다.

# torch.nn
---
torch.nn의 주요 개념

1. `torch.nn.Module` : 신경망의 뻐대(모든 모델의 부모 클래스)
Pytorch로 신경만 모델을 만들 때 반드시 상속받아야하는 기본 클래스.
`class TorchModel(nn.Module)` 과 같이 진행한다.


```python
import torch
import torch.nn as nn
import torch.optim as optim
import matplotlib.pyplot as plt
import matplotlib.font_manager as fm

font_path = r'C:\Windows\Fonts\malgun.ttf'
font_name = fm.FontProperties(fname = font_path).get_name()

plt.rc('font', family = font_name)
plt.rcParams['axes.unicode_minus'] = False

torch.manual_seed(123)

x = torch.linspace(0, 10, 1000).reshape(-1, 1)
noise = torch.randn(x.shape) * 3
y = 3 * x + 10 + noise

class LinearRegressionModel(nn.Module):
    def __init__(self):
        super().__init__()
        self.linear = nn.Linear(1, 1)
    def forward(self, x):
        return self.linear(x)
    
model = LinearRegressionModel()

criterion = nn.MSELoss()
optimizer = optim.SGD(model.parameters(), lr = 0.01)

epochs = 1000
cost_history = []

for epoch in range(epochs + 1):
    hypothesis = model(x)
    cost = criterion(hypothesis, y)
    cost_history.append(cost.item())
    optimizer.zero_grad()
    cost.backward()
    optimizer.step()

    if(epoch + 1) % 100 == 0:
        w_val = model.linear.weight.item()
        b_val = model.linear.bias.item()
        print(f"Epoch : {epoch}, Weight : {w_val:.3f}, Bias : {b_val:.3f}, Cost : {cost.item():.3f}")

fig, axes = plt.subplots(1, 2, figsize = (14, 6))
axes[0].plot(cost_history[:20], color = 'crimson', linewidth = 2, label = "MSE(Cost)")
axes[0].set_title("Epoch에 따른 Cost")
axes[0].set_xlabel("Epoch", fontsize = 10)
axes[0].set_ylabel("Cost", fontsize = 10)
axes[0].grid(True, linestyle='--', alpha = 0.6)
axes[0].legend()

final_w = model.linear.weight.item()
final_b = model.linear.bias.item()

axes[1].scatter(x.numpy(), y.numpy(), color = 'dodgerblue', alpha = 0.5, label = 'Noise Data (3x X 10 + noise)')
axes[1].plot(x.numpy(), 3 * x.numpy() + 10, color = 'green', linestyle = '--', linewidth = 1.5, label = 'actual data')
axes[1].plot(x.numpy(), model(x).detach().numpy(), color = 'red', linewidth = 2, label = f"학습된 weight, bias {final_w:.3f} * x + {final_b:.3f}")
axes[1].set_title('선형 회귀 적합 결과 비교', fontsize = 12)
axes[1].set_xlabel('X', fontsize = 10)
axes[1].set_ylabel('Y', fontsize = 10)
axes[1].grid(True, linestyle = '--', alpha = 0.6)
axes[1].legend()

plt.tight_layout()
plt.show()
```
