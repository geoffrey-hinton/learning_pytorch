```python
import torch

# 1. CUDA 사용 가능 여부 확인
cuda_available = torch.cuda.is_available()
print(f"CUDA 사용 가능 여부: {cuda_available}")

if cuda_available:
    # 2. 현재 GPU 장치 이름 및 개수 출력
    print(f"연결된 GPU 장치 수: {torch.cuda.device_count()}")
    print(f"현재 GPU 모델명: {torch.cuda.get_device_name(0)}")
    
    # 3. 간단한 GPU 텐서 연산 테스트
    x = torch.ones(3, 3, device="cuda")
    print("\n[GPU 텐서 연산 테스트 성공]")
    print(x)
else:
    print("\n[주의] GPU를 인식하지 못하고 CPU 모드로 동작 중입니다.")
```
